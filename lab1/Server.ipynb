# Telefonauskunft: Server

```python
import clientserver

phonebook = clientserver.PhoneBook({
    "Person001": "042-123456",
    "Person002": "042-987654",
    "Person003": "042-555555",
    "Person004": "042-111111",
    "Person005": "042-222222",
    "Person006": "042-333333",
})

server = clientserver.Server(phonebook)
server.serve()
```

```python
import clientserver

phonebook = clientserver.PhoneBook.with_random_entries(500, seed=42)
server = clientserver.Server(phonebook)
server.serve()
```
